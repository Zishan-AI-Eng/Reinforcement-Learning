```python
import numpy as np
np.bool8 = np.bool_  
import gym
import cv2
```

```python
ROWS = 4
COLS = 12
CELL_SIZE = 60

def initialize_frame():
    """
    Creates the base grid image for CliffWalking-v0.
    Colors: White (Safe), Dark Gray (Cliff), Green (Start), Gold (Goal).
    """
    
    img = np.ones((ROWS * CELL_SIZE, COLS * CELL_SIZE, 3), dtype=np.uint8) * 255

    for r in range(ROWS):
        for c in range(COLS):
            state = r * COLS + c
            top_left = (c * CELL_SIZE, r * CELL_SIZE)
            bottom_right = ((c + 1) * CELL_SIZE, (r + 1) * CELL_SIZE)

            if state == 36:     
                color = (144, 238, 144) 
            elif state == 47:   
                color = (255, 215, 0)  
            elif 37 <= state <= 46: 
                color = (100, 100, 100)  
            else:
                color = (255, 255, 255) 
                
            cv2.rectangle(img, top_left, bottom_right, color, -1)
            
            cv2.rectangle(img, top_left, bottom_right, (0, 0, 0), 1)

    return img


def put_agent(img, state):
    """
    Takes the base frame and draws a blue circle representing the agent at the current state.
    """
    frame = img.copy()
    
    r = state // COLS
    c = state % COLS
    center_x = int(c * CELL_SIZE + CELL_SIZE / 2)
    center_y = int(r * CELL_SIZE + CELL_SIZE / 2)
    radius = int(CELL_SIZE / 3)

    cv2.circle(frame, (center_x, center_y), radius, (0, 0, 255), -1)
    
    cv2.circle(frame, (center_x, center_y), radius, (0, 0, 0), 2)

    return frame
```

```python
cliffEnv = gym.make("CliffWalking-v0")
done = False
frame = initialize_frame()


reset_result = cliffEnv.reset()
state = reset_result[0] if isinstance(reset_result, tuple) else reset_result

while not done:
    frame2 = put_agent(frame.copy(), state)
    cv2.imshow('CliffWalking-v0', frame2)
    cv2.waitKey(250)
    action = int(np.random.randint(low=0, high=4))
    print(state, '--->' , ['Up', 'Right', 'Down', 'Left'][action])
    
    
    step_result = cliffEnv.step(action)
    state = step_result[0]
    
    
    if len(step_result) == 5:
        done = step_result[2] or step_result[3]
    else:
        done = step_result[2]

cliffEnv.close()
```
