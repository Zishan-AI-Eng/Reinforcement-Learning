```python
import numpy as np 
import gym
import pickle as pkl
import cv2
```

```python

# ==========================================
# PARAMETERS
# ==========================================
EPSILON = 0.1
ALPHA = 0.1
GAMMA = 0.9
NUM_EPISODE = 500
```

```python
cliffEnv= gym.make('CliffWalking-v0')
q_table = np.zeros(shape=(48, 4))


def policy(state, explore=0.0):
    if np.random.random() <= explore:
        return int(np.random.randint(0, q_table.shape[1]))

    return int(np.argmax(q_table[state]))
        
```

```python
for episode in range(NUM_EPISODE):
    done = False
    total_reward = 0
    episode_length = 0

    # Gym reset ab tuple (state, info) return karta hai
    state, info = cliffEnv.reset()

    while not done:
        # 1. State ke hisab se action select karein
        action = policy(state, EPSILON)

        # 2. Environment mein step lein
        next_state, reward, terminated, truncated, info = cliffEnv.step(action)
        done = terminated or truncated

        # 3. Target calculate karein (The Q-Learning Magic)
        if done:
            target = reward
        else:
            target = reward + GAMMA * np.max(q_table[next_state])

        # 4. Q-Table update karein
        q_table[state, action] += ALPHA * (target - q_table[state, action])

        # 5. Agle step ke liye state update karein
        state = next_state
        
        # 6. Metrics update karein
        total_reward += reward
        episode_length += 1

    # Har episode ke baad print karein
    print(
        f"Episode --> {episode + 1} | "
        f"Episode Length --> {episode_length} | "
        f"Total Reward --> {total_reward}"
    )

cliffEnv.close()
```

```python
with open("q_learning_q_table.pkl", "wb") as f:
    pkl.dump(q_table, f)
    
print("Training Complete! Q-table saved as 'q_learning_q_table.pkl'.")
```
