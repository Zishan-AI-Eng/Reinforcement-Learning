```python
import numpy as np
import gym
import pickle as pkl
```

```python

# ==========================================
# HYPERPARAMETERS
# ==========================================
EPSILON = 0.1
ALPHA = 0.1
GAMMA = 0.9
NUM_EPISODE = 500
```

```python

# ==========================================
# ENVIRONMENT & Q-TABLE SETUP
# ==========================================
cliffEnv = gym.make("CliffWalking-v0")

# Q-table ko zeros se initialize karein (48 states, 4 actions)
q_table = np.zeros((cliffEnv.observation_space.n, cliffEnv.action_space.n))

def policy(state, explore=0.0):
    if np.random.random() <= explore:
        return int(np.random.randint(0, q_table.shape[1]))
    return int(np.argmax(q_table[state]))
```

```python

# ==========================================
# Q-LEARNING TRAINING LOOP
# ==========================================
print("Training Started...")

for episode in range(NUM_EPISODE):
    done = False
    total_reward = 0
    episode_length = 0

    # Gym reset ab tuple (state, info) return karta hai
    state, info = cliffEnv.reset()

    while not done:
        # 1. State ke hisab se action select karein (Epsilon-greedy)
        action = policy(state, EPSILON)

        # 2. Environment mein step lein
        next_state, reward, terminated, truncated, info = cliffEnv.step(action)
        done = terminated or truncated

        # 3. Target calculate karein (The Q-Learning Magic using np.max)
        if done:
            target = reward
        else:
            target = reward + GAMMA * np.max(q_table[next_state])

        # 4. Q-Table update karein
        q_table[state, action] += ALPHA * (target - q_table[state, action])

        # 5. Agle step ke liye state update karein
        state = next_state
        
        # 6. Metrics update karein
        total_reward += reward
        episode_length += 1

    # Print progress (har 50 episodes ke baad taake console saaf rahay)
    if (episode + 1) % 50 == 0 or episode == 0:
        print(
            f"Episode --> {episode + 1:3d} | "
            f"Episode Length --> {episode_length:3d} | "
            f"Total Reward --> {total_reward}"
        )

cliffEnv.close()
```

```python

# ==========================================
# SAVE THE TRAINED MODEL
# ==========================================
with open("q_learning_q_table.pkl", "wb") as f:
    pkl.dump(q_table, f)
    
print("\nTraining Complete! Q-table saved as 'q_learning_q_table.pkl'.")
```
